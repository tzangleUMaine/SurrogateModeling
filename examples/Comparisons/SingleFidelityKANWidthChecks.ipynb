# Single Fidelity KAN Width Checks

Tests network `width` (capacity) as a separate lever from grid resolution and regularization, kept in its own notebook so it doesn't clutter `SingleFidelityKANSpotChecks.ipynb`.

Baseline config for each dimension is the current best-known setup from the spot checks: `grid=[3, 5]`, `lamb=0`, `n=1000`, `n_iter=400`. Each dimension here reruns that baseline width alongside a narrower (0.5x hidden layers) and wider (2x hidden layers) variant, with depth and everything else held fixed, so width is the only thing changing. Unlike grid refinement, a width change is a structurally different network with no warm start -- every variant here is a full-cost retrain.

Starting with 4D only; add other dimensions here the same way once this is validated.

```python
%load_ext autoreload
%autoreload 2

import numpy as np
import matplotlib.pyplot as plt

import examples.multi_fidelity_functions as mfs
from examples import single_fidelity_kan_helpers as kan_helpers
```

```python
key = "4D"
my_function = mfs.increasing_dimensions[key]()
X_test, y_test = kan_helpers.create_test_points(my_function, n=1000)
```

### 4D Seed-Variance Check: width [4, 6, 6, 1] vs [4, 8, 8, 1]

`[4,6,6,1]` came out ~9x better than `[4,8,8,1]` on a single seed (`seed=0`), which is a bigger gap than I'd trust from one run each -- width changes the whole random initialization, and LBFGS on a non-convex loss can land in very different quality optima depending on init. Running two more seeds per width to see whether this ordering holds up or was a lucky/unlucky draw. Reuses the existing `seed=0` results from above rather than re-running them.

```python
models_w66_seeds = [(0, error_4d_w66, model_4d_w66)]
for seed in [1, 2]:
    m, err, tt, et = kan_helpers.generate_kan_model(
        name=key,
        func=my_function,
        n=1000,
        seed=seed,
        X_test=X_test,
        y_test=y_test,
        width=[4, 6, 6, 1],
        grid=[3, 5],
        spline_order=3,
        opt="LBFGS",
        n_iter=400,
        lr=1.0,
        lamb=0.0,
    )
    models_w66_seeds.append((seed, err, m))
    print(f"width=[4,6,6,1] seed={seed} -> model_error={err}, train_time={tt:.1f}s")
    kan_helpers.print_convergence_report(f"width=[4,6,6,1] seed={seed}", m.train_losses, m.test_losses)
```

```python
models_w88_seeds = [(0, error_4d_baseline, model_4d_baseline)]
for seed in [1, 2]:
    m, err, tt, et = kan_helpers.generate_kan_model(
        name=key,
        func=my_function,
        n=1000,
        seed=seed,
        X_test=X_test,
        y_test=y_test,
        width=[4, 8, 8, 1],
        grid=[3, 5],
        spline_order=3,
        opt="LBFGS",
        n_iter=400,
        lr=1.0,
        lamb=0.0,
    )
    models_w88_seeds.append((seed, err, m))
    print(f"width=[4,8,8,1] seed={seed} -> model_error={err}, train_time={tt:.1f}s")
    kan_helpers.print_convergence_report(f"width=[4,8,8,1] seed={seed}", m.train_losses, m.test_losses)
```

```python
errs_66 = [e for _, e, _ in models_w66_seeds]
errs_88 = [e for _, e, _ in models_w88_seeds]
print(f"width=[4,6,6,1] errors over seeds 0,1,2: {errs_66}")
print(f"  -> mean={np.mean(errs_66):.4g}, std={np.std(errs_66):.4g}")
print(f"width=[4,8,8,1] errors over seeds 0,1,2: {errs_88}")
print(f"  -> mean={np.mean(errs_88):.4g}, std={np.std(errs_88):.4g}")
```

### 4D -- is the grid=5 refinement worth it for width [4, 4, 4, 1]?

The `grid=[3, 5]` run for `[4, 4, 4, 1]` showed test error barely moving (and slightly worsening) between the end of the grid=3 stage (epoch 399: test=0.002056) and the end of training (epoch 799, after refining to grid=5: test=0.002189) -- the refinement bought train-error improvement but no test-error benefit. Testing `grid=[3]` alone, same `n_iter=400`, to see if dropping the refinement matches the grid=[3,5] result at roughly half the training cost.

```python
model_4d_grid3only, error_4d_grid3only, train_time_4d_grid3only, eval_time_4d_grid3only = kan_helpers.generate_kan_model(
    name=key,
    func=my_function,
    n=1000,
    seed=0,
    X_test=X_test,
    y_test=y_test,
    width=[4, 4, 4, 1],
    grid=[3],
    spline_order=3,
    opt="LBFGS",
    n_iter=400,
    lr=1.0,
    lamb=0.0,
)
```

```python
ax = model_4d_grid3only.plot_fit_error("4D KAN Fit Error n=1000, width=[4,4,4,1], grid=[3] only")
plt.yscale("log")
kan_helpers.save_plot(ax, "4D_grid3only_lamb0_width4-4-4-1")
plt.show()
print(f"4D (width [4,4,4,1], grid=[3] only) model_error = {error_4d_grid3only}")
```

```python
print(f"4D [4,4,4,1] grid=[3,5] -> model_error={error_4d_narrow}, train_time={train_time_4d_narrow:.1f}s")
print(f"4D [4,4,4,1] grid=[3]   -> model_error={error_4d_grid3only}, train_time={train_time_4d_grid3only:.1f}s")

kan_helpers.print_convergence_report("4D [4,4,4,1] grid=[3,5]", model_4d_narrow.train_losses, model_4d_narrow.test_losses)
kan_helpers.print_convergence_report("4D [4,4,4,1] grid=[3]", model_4d_grid3only.train_losses, model_4d_grid3only.test_losses)
```
